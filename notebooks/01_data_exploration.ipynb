```python
import re
import nltk

nltk.download('stopwords')

from nltk.corpus import stopwords
```

```python
def clean_text(text):

    text = text.lower()

    text = re.sub(r'<.*?>', '', text)

    text = re.sub(r'http\S+|www\S+', '', text)

    text = re.sub(r'[^a-zA-Z\s]', '', text)

    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score
from sklearn.utils.class_weight import compute_sample_weight

X_train_sub, X_val, y_train_sub, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train
)

# Calculate weights using only the new training subset
val_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_sub
)

alpha_values = [0.01, 0.1, 0.5, 1.0, 2.0, 5.0, 10.0]

for alpha in alpha_values:

    model = MultinomialNB(alpha=alpha)

    model.fit(
        X_train_sub,
        y_train_sub,
        sample_weight=val_weights
    )

    val_predictions = model.predict(X_val)

    accuracy = accuracy_score(y_val, val_predictions)

    macro_f1 = f1_score(
        y_val,
        val_predictions,
        average="macro"
    )

    print(f"\nAlpha: {alpha}")
    print("Accuracy:", accuracy)
    print("Macro F1-score:", macro_f1)
```

```python
from sklearn.model_selection import train_test_split

text_train, text_test, sentiment_train, sentiment_test = train_test_split(
    df_clean["processed_review"],
    df_clean["sentiment"],
    test_size=0.2,
    random_state=42,
    stratify=df_clean["sentiment"]
)

print("Training samples:", len(text_train))
print("Testing samples:", len(text_test))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

lr_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(max_features=5000)),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

print("Machine Learning pipeline created successfully!")
```

```python
param_grid = {
    "classifier__C": [0.1, 1, 10],
    "tfidf__ngram_range": [(1, 1), (1, 2)]
}

print("Hyperparameter combinations:")
print(param_grid)
```

```python
from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=param_grid,
    cv=3,
    scoring="f1_macro",
    n_jobs=-1,
    verbose=1
)

print("GridSearchCV created successfully!")
```

```python
grid_search.fit(text_train, sentiment_train)

print("Hyperparameter tuning completed!")
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation Macro F1:")
print(grid_search.best_score_)
```

```python
best_lr_model = grid_search.best_estimator_

y_pred_optimized = best_lr_model.predict(text_test)

print("Optimized Logistic Regression Accuracy:")
print(accuracy_score(sentiment_test, y_pred_optimized))

print("\nOptimized Classification Report:")
print(classification_report(sentiment_test, y_pred_optimized))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, f1_score

nb_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 2)
    )),
    ("classifier", MultinomialNB(alpha=0.1))
])

nb_pipeline.fit(text_train, sentiment_train)

y_pred_nb = nb_pipeline.predict(text_test)

print("Naive Bayes Accuracy:", accuracy_score(sentiment_test, y_pred_nb))

print("\nNaive Bayes Macro F1:",
      f1_score(sentiment_test, y_pred_nb, average="macro"))

print("\nClassification Report:")
print(classification_report(sentiment_test, y_pred_nb))
```

```python
from sklearn.linear_model import LogisticRegression

lr_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 2)
    )),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

lr_pipeline.fit(text_train, sentiment_train)

y_pred_lr = lr_pipeline.predict(text_test)

print("Logistic Regression Accuracy:",
      accuracy_score(sentiment_test, y_pred_lr))

print("Logistic Regression Macro F1:",
      f1_score(sentiment_test, y_pred_lr, average="macro"))

print("\nClassification Report:")
print(classification_report(sentiment_test, y_pred_lr))
```

```python
from sklearn.ensemble import VotingClassifier

voting_pipeline = VotingClassifier(
    estimators=[
        ("nb", nb_pipeline),
        ("lr", lr_pipeline)
    ],
    voting="soft"
)

voting_pipeline.fit(text_train, sentiment_train)

y_pred_voting = voting_pipeline.predict(text_test)

print("Soft Voting Accuracy:",
      accuracy_score(sentiment_test, y_pred_voting))

print("Soft Voting Macro F1:",
      f1_score(sentiment_test, y_pred_voting, average="macro"))

print("\nClassification Report:")
print(classification_report(sentiment_test, y_pred_voting))
```

```python
model_comparison = pd.DataFrame({
    "Model": [
        "Naive Bayes",
        "Logistic Regression",
        "Soft Voting Ensemble"
    ],
    "Accuracy": [
        accuracy_score(sentiment_test, y_pred_nb),
        accuracy_score(sentiment_test, y_pred_lr),
        accuracy_score(sentiment_test, y_pred_voting)
    ],
    "Macro F1": [
        f1_score(sentiment_test, y_pred_nb, average="macro"),
        f1_score(sentiment_test, y_pred_lr, average="macro"),
        f1_score(sentiment_test, y_pred_voting, average="macro")
    ]
})

display(model_comparison.round(4))
```
